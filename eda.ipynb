```python
sns.heatmap(df.isnull(), cbar=False, cmap="viridis", yticklabels=False)
plt.tight_layout()
plt.savefig("images/null_heatmap.png")
plt.show()
```

>> Missingness is isolated

```python
plt.figure(figsize=(12, 12))
sns.heatmap(df.drop(columns=["ocean_proximity"]).corr(), annot=True, linecolor="white")
plt.tight_layout()
plt.savefig("images/corr_map.png")
plt.show()
```

```python
numeric_cols = [
    "longitude",
    "latitude",
    "housing_median_age",
    "total_rooms",
    "total_bedrooms",
    "population",
    "households",
    "median_income",
    "median_house_value",
]
```

```python
fig, axes = plt.subplots(3, 3, figsize=(15, 12))
axes = axes.flatten()

for i, col in enumerate(numeric_cols):
    sns.boxplot(y=df[col], ax=axes[i])
    axes[i].set_title(col)

plt.tight_layout()
plt.savefig("images/boxplots.png")
plt.show()
```

```python
fig, axes = plt.subplots(3, 3, figsize=(15, 12))
axes = axes.flatten()

for i, col in enumerate(numeric_cols):
    sns.kdeplot(x=df[col], ax=axes[i])
    axes[i].set_title(col)

plt.tight_layout()
plt.savefig("images/kdeplots.png")
plt.show()
```
