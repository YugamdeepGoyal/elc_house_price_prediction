```python
# Set the figure size using matplotlib
plt.figure(figsize=(9, 7))

# Create the scatter plot using seaborn
scatter = sns.scatterplot(
    data=df,
    x="longitude",
    y="latitude",
    hue="median_house_value",
    palette="viridis",
    size="population",
    sizes=(20, 200),  # Adjusts the minimum and maximum bubble sizes
    alpha=0.4,
    legend="brief"    # Shows a clean, binned legend for both color and size
)

# Customize labels and title
plt.xlabel("Longitude")
plt.ylabel("Latitude")
plt.title("Geographic Distribution of Median House Value")

# Adjust legend position if it overlaps with data points
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()
plt.savefig("images/geo_distribution_of_median_value.png", dpi=600)
plt.show()
```
